```python
from ultralytics import YOLO

model = YOLO(r"C:\Users\user\Downloads\runs\detect\train-23\weights\best.pt")


metrics = model.val(
    data=r"C:\Users\user\Downloads\Bahraini-Food-Detection-1\data.yaml",
    split="val",
    plots=True
)

print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```
